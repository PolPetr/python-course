```python
pipeline_config = {
    "steps": {
        "tokenization": {"enabled": True, "method": "word"},
        "stopwords": {"enabled": True, "language": "english", "custom_words": []},
        "stemming": {"enabled": False, "algorithm": "porter"},
        "normalization": {"enabled": True, "lowercase": True, "remove_punct": True}
    },
    "input_encoding": "utf-8",
    "output_format": "tokens"
}
#Включите stemming, установив "enabled": True
pipeline_config["steps"]["stemming"]["enabled"] = True
#Добавьте "numbers" в custom_words для стоп-сло
pipeline_config["steps"]["stopwords"]["custom_words"] = "numbers"
#Получите список всех включенных шагов пайплайна
enabled_steps = []
for step_name in pipeline_config["steps"].items():
  if pipeline_config.get("enabled", False):
    enabled_steps.append(step_name)
    print(f"Список всех включенных шагов: {enabled_steps}")
#Измените output_format на "vectors"
pipeline_config["output_format"] = "vectors"
#Создайте упрощенную конфигурацию только с включенными шагами
simplified_config = {
    "input_encoding": pipeline_config["input_encoding"],
    "output_format": pipeline_config["output_format"],
    "steps":
    {step_name: step_config
      for step_name, step_config in pipeline_config["steps"].items()
      if step_config.get("enabled", False)}
}
```

```python
models_stats = {
    "bert-base": {
        "accuracy": 0.92,
        "f1_score": 0.91,
        "inference_time": 120,
        "size_mb": 440
    },
    "distilbert": {
        "accuracy": 0.89,
        "f1_score": 0.88,
        "inference_time": 65,
        "size_mb": 250
    },
    "roberta-large": {
        "accuracy": 0.94,
        "f1_score": 0.93,
        "inference_time": 210,
        "size_mb": 1600
    }
}
#Найдите модель с максимальной точностью (accuracy)
model_max_accuracy = max(models_stats.items(), key=lambda x: x[1]["accuracy"])
print(f"Модель с максимальной точностью: {model_max_accuracy}")
#Рассчитайте среднее время инференса по всем моделям
average_inference_time = (models_stats["bert-base"]["inference_time"] + models_stats["distilbert"]["inference_time"] + models_stats["roberta-large"]["inference_time"]) / 3
print(f"Среднее время инференса по всем моделям: {average_inference_time}")
#Создайте новый словарь только с метриками accuracy и f1_score для каждой модели
new_dict = {
    model_name: {
    "accuracy": statistics["accuracy"],
    "f1_score": statistics["f1_score"]
    }
    for model_name, statistics in models_stats.items()
}
#Добавьте новую модель "albert-base" с данными: accuracy=0.87, f1_score=0.86, inference_time=55, size_mb=180
models_stats["albert-base"]= {
    "accuracy": 0.87,
    "f1_score": 0.86,
    "inference_time": 55,
    "size_mb": 180
}
#Отфильтруйте модели, размер которых меньше 500 МБ
filtered_models = {
    model_name: statistics
    for model_name, statistics in models_stats.items()
    if statistics["size_mb"] < 500
}
print(f"Модели размером меньше 500 МБ: {filtered_models}")
```
